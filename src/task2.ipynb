```python
import tensorflow as tf
import numpy as np
```

```python
# Input data
x = np.array([
    [1, 1],
    [2, 1],
    [3, 2],
    [4, 3]
], dtype=float)
```

```python
# Output
y = np.array([3, 5, 8, 11], dtype=float)
```

```python
# Model
model = tf.keras.Sequential([
    # Hidden layer
    tf.keras.layers.Dense(5, activation='relu', input_shape=[2]),
    # Output layer
    tf.keras.layers.Dense(1)
])
```

```python
# Compile
model.compile(
    optimizer='sgd',
    loss='mean_squared_error'
)
```

```python
# Train
model.fit(x, y, epochs=500)
```

```python
# Predict
prediction = model.predict(np.array([[5, 4]]))
print(prediction)
```
