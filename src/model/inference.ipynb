# Build real inference engine using TensorFlow Lite

```python
import tensorflow as tf
import numpy as np
```

```python
# Load TFLite model
interpreter = tf.lite.Interpreter(
    model_path="mnist_model.tflite"
)
```

```python
# Allocate memory
interpreter.allocate_tensors()
```

```python
# Input details
input_details = interpreter.get_input_details()
```

```python
# Output details
output_details = interpreter.get_output_details()
```

```python
print("Input Details:")
print(input_details)

print("\nOutput Details:")
print(output_details)
```

# REAL inference part

```python
from tensorflow import keras

# Load MNIST
(_, _), (x_test, y_test) = keras.datasets.mnist.load_data()

# Normalize
x_test = x_test / 255.0
```

```python
# Select one image
image = x_test[2]
```

```python
# Add batch dimension
input_data = np.array([image], dtype=np.float32)
```

```python
# Set input tensor
interpreter.set_tensor(
    input_details[0]['index'],
    input_data
)
```

```python
# Run inference
interpreter.invoke()
```

```python
# Get output
output_data = interpreter.get_tensor(
    output_details[0]['index']
)
```

```python
print("Raw Output:")
print(output_data)
```

```python
# Predicted digit
predicted_digit = np.argmax(output_data)
print("Predicted Digit:", predicted_digit)
print("Real Digit:", y_test[2])
```
